### Q15 -REMOVE TEXT FROM RUNTIME AND CONVERT IT TO INTEGER :

```python
df['Runtime'] = df['Runtime'].str.replace(' min', '', regex=False).astype(int)
print(df['Runtime'])
print(df['Runtime'].dtype)
```

### Q16- FILL MISSING VALUES IN GROSS WITH THE MEDIAN :

```python
df['Gross'] = df['Gross'].fillna(df['Gross'].median())
print(df['Gross'].isnull().sum())
```

### Q17- COUNT MOVIES IN EACH CERTIFICATE CATEGORY :

```python
print (df['Certificate'].value_counts())
```

### Q18- CORRELATION BETWEEN IMDB_RATING AND META_SCORE :

```python
correlation = df['IMDB_Rating'].corr(df['Meta_score'])
print ("Correlation coefficient: ", correlation)
```

### Q19- CREATE RATING 100 SCALE :

```python
df["Rating_100_scale"] = df["IMDB_Rating"]*10
print(df[['IMDB_Rating', 'Rating_100_scale']].head())
```

### Q20- COUNT MOVIES RELEASED FROM 1990 TO 2000 :

```python
df['Released_Year'] = pd.to_numeric(df['Released_Year'] , errors = "coerce")
movies_1990_2000 = df[
    (df['Released_Year'] >= 1990) &
    (df['Released_Year'] <= 2000)]

print("Total movies released from 1990 to 2000:", len(movies_1990_2000))

print(movies_1990_2000)
```

### Q21- HISTOGRAM OF IMDB RATINGS : 

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(9, 6))
plt.hist(df['IMDB_Rating'].dropna(), bins=15, edgecolor='black')

plt.title('Distribution of IMDb Ratings')
plt.xlabel('IMDb Rating')
plt.ylabel('Number of Movies')
plt.tight_layout()
plt.show()
```

### Q22- BAR CHART OF THE TOP 10 DIRECTORS :

```python
top_directors = df['Director'].value_counts().head(10)

plt.figure(figsize=(12, 8))
top_directors.plot(kind='bar', edgecolor='black')

plt.title('Top 10 Directors by Number of Movies')
plt.xlabel('Director')
plt.ylabel('Number of Movies')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

### Q23- SCATTER PLOT OF IMDB RATING VS GROSS REVENUE :

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df['IMDB_Rating'],
    df['Gross'],
    alpha=0.6
)

plt.title('IMDb Rating vs Gross Revenue')
plt.xlabel('IMDb Rating')
plt.ylabel('Gross Revenue')
plt.tight_layout()
plt.show()
```

### Q24- PIE CHART OF THE TOP 5 GENRES :

```python
top_genres = df['Genre'].value_counts().head(5)

plt.figure(figsize=(8, 8))
plt.pie(
    top_genres,
    labels=top_genres.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Top 5 Genres by Movie Count')
plt.tight_layout()
plt.show()
```

### Q25- BOX PLOT OF META SCORE :

```python
plt.figure(figsize=(8, 5))

plt.boxplot(df['Meta_score'].dropna())

plt.title('Box Plot of Meta Score')
plt.ylabel('Meta Score')
plt.tight_layout()
plt.show()
```

### Q26-BAR PLOT OF AVERAGE GROSS BY CERTIFICATE :

```python
avg_gross = df.groupby('Certificate')['Gross'].mean().dropna()
avg_gross.plot(kind='bar', figsize=(10, 5), edgecolor='black')

plt.title('Average Gross Revenue by Certificate')
plt.xlabel('Certificate')
plt.ylabel('Average Gross Revenue')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Q27- LINE PLOT OF MOVIES RELEASED EACH YEAR : 

```python
df['Released_Year'] = pd.to_numeric(
    df['Released_Year'], errors='coerce')

year_counts = df['Released_Year'].value_counts().sort_index()

plt.figure(figsize=(12, 5))
plt.plot(year_counts.index, year_counts.values, marker='o')

plt.title('Number of Movies Released Each Year')
plt.xlabel('Release Year')
plt.ylabel('Number of Movies')
plt.grid(True)
plt.tight_layout()
plt.show()
```

### Q28- HORIZONTAL BAR CHART OF THE TOP 5 HIGHEST - VOTED MOVIES :

```python
top_5_movies = df.nlargest(5, 'No_of_Votes')

plt.figure(figsize=(10, 6))
plt.barh(top_5_movies['Series_Title'], top_5_movies['No_of_Votes'])

plt.title('Top 5 Highest-Voted Movies')
plt.xlabel('Number of Votes')
plt.ylabel('Movie Title')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

### Q29- COUNT UNIQUE GENRES IN THE DATASET :

### Q36- COMPARE GROSS REVENUE FOR DRAMA AND ACTION MOVIES :

```python
import matplotlib.pyplot as plt

drama = df[
    df['Genre'].str.contains('Drama', na = False)
]['Gross'].dropna()

action = df[
    df['Genre'].str.contains('Action', na = False)
]['Gross'].dropna()

plt.figure(figsize=(8, 5))
plt.boxplot([drama, action], tick_labels=['Drama', 'Action'])

plt.title('Gross Revenue: Drama vs Action')
plt.xlabel('Genre')
plt.ylabel('Gross Revenue')
plt.tight_layout()
plt.show()
```

### Q39- ACTORS WHO ARE ALSO DIRECTORS :

### Q40- COMPARE META_SCORE AND IMDB_RATING FOR THE HIGHEST RATED MOVIES :

```python
top_movies = df.nlargest(10, 'IMDB_Rating').copy()

top_movies['IMDB_Rating_Scaled'] = top_movies['IMDB_Rating'] * 10

top_movies.plot(x='Series_Title',y=['Meta_score' , 'IMDB_Rating_Scaled'] , kind='bar' , figsize=(12, 6))

plt.title('Meta Score vs IMDb Rating (Scaled to 100)')
plt.xlabel('Movie')
plt.ylabel('Score out of 100')
plt.xticks(rotation=45, ha='right')
plt.legend(['Meta Score', 'IMDb Rating × 10'])
plt.tight_layout()
plt.show()
```
